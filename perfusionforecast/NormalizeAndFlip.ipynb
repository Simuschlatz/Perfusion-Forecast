```python
import os
import numpy as np
from rich.progress import track
from pathlib import Path
```

```python
source=Path("../QualityFiltered")
destination=Path("../NormalizedBoth")
for file in track(source.glob("*.npy")):
    volume_seq = np.load(file)
    volume_seq = (volume_seq - volume_seq.min())/(volume_seq.max() - volume_seq.min())
    volume_seq_flipped = np.flip(volume_seq, axis=-1)
    target_path = destination / file.name
    flipped_name = file.stem + "_flipped" + file.suffix
    target_path_flipped = destination / flipped_name
    np.save(target_path, volume_seq)
    np.save(target_path_flipped, volume_seq_flipped)
```
